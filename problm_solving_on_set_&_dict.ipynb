19. Count Frequency of Elements

```python
def freq(lst):
  res={}
  for num in lst:
    res.update({num:lst.count(num)})
  return res
print(freq([1,2,2,3]))

```

25. Combine Values of Duplicate Keys

```python
d1 = {"a": 1, "b": 2}
d2 = {"a": 3, "c": 4}
res = d1.copy()
for key in d2:
    if key in res:
        res[key] = res[key] + d2[key]
    else:
        res[key] = d2[key]
print(res)
```

26. Count Word Frequency in Sentence

```python
n="apple banana apple"
sen=n.split()
d={}
for i in sen:
  if i in d:
    d[i]+=1
  else:
    d[i]=1
print(d)
```

27. Remove Duplicate Values from Dictionary

```python
n = {"a": 1, "b": 2, "c": 1}
res = {}
for key, value in n.items():
    if value not in res.values():
        res[key] = value
print(res)
```

28. Find Common Keys in Two Dictionaries

```python
d1={"a":1,"b":2}
d2={"b":3,"c":4}
res=[]
for key in d1:
  if key in d2:
    res.append(key)
print(res)
```

29. Swap Keys and Values Safely

```python
n={"x":1,"y":2}
res={v:k for k,v in n.items()}
print(res)
```

30. Delete Items by Value

```python
n = {"a": 1, "b": 2, "c": 1}
res = {}
for key, value in n.items():
    if value != 1:
        res[key] = value
print(res)
```
